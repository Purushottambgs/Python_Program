```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris= pd.read_csv('pandas/Dataset/iris.csv')
iris.sample(5)
```

```python
iris['Species']= iris['Species'].replace({'Iris-versicolor':0, 'Iris-virginica':1, 'Iris-setosa':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.7)
plt.xlabel('Sepal length')
plt.ylabel('Patal Length')
plt.colorbar()
```

```python
plt.figure(figsize=(15,7)) # size change karne ke liye

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='summer', alpha=0.7) # cmap basically it is thems
plt.title('it is a iris plot')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
batters=pd.read_csv('pandas/Dataset/batter.csv')
batters.head(5)
```

```python
sample_df=batters.head(100).sample(25)
sample_df.shape
sample_df
```

```python
plt.figure(figsize=(12,8))
plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])

for i in range (sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
# horizontal and vertical lines

plt.figure(figsize=(12,8))
plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])
plt.axhline(130, color='red')
plt.axvline(30, color='red')

for i in range (sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
fig, ax= plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10,6))
ax[0].scatter(batters['avg'], batters['strike_rate'], color='red')
ax[1].scatter(batters['avg'], batters['runs'], color='yellow')

ax[0].set_title('avg vs strike rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```
